```python
# ============================================================================
# === CELL 1: Setup and Configuration ===
# Run this first to set up the environment
# ============================================================================

import pandas as pd
import numpy as np
import os
import math
import gc
import time
from datetime import datetime

# ----- CONFIGURATION - Adjust these values as needed -----
BATCH_SIZE = 5000           # Participants per batch (reduce if memory issues)
CHUNK_SIZE = 500            # Person IDs per SQL query (reduce if query fails)
RANDOM_STATE = 42           # For reproducibility

# Output directories
OUTPUT_DIR = "merged_df"
BATCH_DIR = f"{OUTPUT_DIR}/batches"
RESULTS_DIR = f"{OUTPUT_DIR}/results"

# Create directories
for d in [OUTPUT_DIR, BATCH_DIR, RESULTS_DIR]:
    os.makedirs(d, exist_ok=True)

# Get CDR (Clinical Data Repository) from environment
CDR = os.environ["WORKSPACE_CDR"]

print("=" * 60)
print("BATCH PROCESSING SETUP COMPLETE")
print("=" * 60)
print(f"CDR: {CDR}")
print(f"Batch size: {BATCH_SIZE}")
print(f"Chunk size: {CHUNK_SIZE}")
print(f"Output directory: {OUTPUT_DIR}")
```

```python
# ============================================================================
# === CELL 5: Define the Batch Processing Function ===
# Core function that processes one batch at a time
# ============================================================================

def process_batch(batch_number, person_ids, cdr):
    """
    Process a single batch of participants.

    Returns:
        tuple: (demographics_df, conditions_df, success_flag, error_message)
    """
    batch_start = time.time()

    # Process in smaller chunks to avoid SQL query limits
    all_demographics = []
    all_conditions = []

    n_chunks = math.ceil(len(person_ids) / CHUNK_SIZE)

    for chunk_idx in range(n_chunks):
        start_idx = chunk_idx * CHUNK_SIZE
        end_idx = min(start_idx + CHUNK_SIZE, len(person_ids))
        chunk_ids = person_ids[start_idx:end_idx]
        chunk_str = ", ".join(str(pid) for pid in chunk_ids)

        try:
            # Demographics query
            demo_sql = f"""
            SELECT
                p.person_id,
                p.year_of_birth,
                p.gender_concept_id,
                g.concept_name as gender,
                p.race_concept_id,
                r.concept_name as race,
                p.ethnicity_concept_id,
                e.concept_name as ethnicity,
                EXTRACT(YEAR FROM CURRENT_DATE()) - p.year_of_birth as age
            FROM `{cdr}.person` p
            LEFT JOIN `{cdr}.concept` g ON p.gender_concept_id = g.concept_id
            LEFT JOIN `{cdr}.concept` r ON p.race_concept_id = r.concept_id
            LEFT JOIN `{cdr}.concept` e ON p.ethnicity_concept_id = e.concept_id
            WHERE p.person_id IN ({chunk_str})
            """

            chunk_demo = pd.read_gbq(demo_sql, dialect='standard')
            all_demographics.append(chunk_demo)

            # Conditions query
            cond_sql = f"""
            SELECT
                co.person_id,
                co.condition_concept_id,
                c.concept_name as condition_name,
                co.condition_start_datetime,
                co.condition_end_datetime
            FROM `{cdr}.condition_occurrence` co
            LEFT JOIN `{cdr}.concept` c ON co.condition_concept_id = c.concept_id
            WHERE co.person_id IN ({chunk_str})
              AND co.condition_concept_id IN ({ALL_CONDITION_IDS_STR})
            """

            chunk_cond = pd.read_gbq(cond_sql, dialect='standard')
            all_conditions.append(chunk_cond)

        except Exception as e:
            return None, None, False, f"Chunk {chunk_idx} failed: {str(e)}"

    # Combine all chunks
    demographics_df = pd.concat(all_demographics, ignore_index=True) if all_demographics else pd.DataFrame()
    conditions_df = pd.concat(all_conditions, ignore_index=True) if all_conditions else pd.DataFrame()

    elapsed = time.time() - batch_start
    print(f"  Batch {batch_number} completed in {elapsed:.1f}s "
          f"({len(demographics_df)} persons, {len(conditions_df)} conditions)")

    return demographics_df, conditions_df, True, None


def create_condition_flags(conditions_df, person_ids):
    """
    Create binary flags for each condition of interest.
    """
    flags_df = pd.DataFrame({'person_id': person_ids})

    for condition_name, concept_ids in CONDITION_CONCEPTS.items():
        col_name = f"has_{condition_name}"

        if len(conditions_df) > 0:
            persons_with_condition = conditions_df[
                conditions_df['condition_concept_id'].isin(concept_ids)
            ]['person_id'].unique()
            flags_df[col_name] = flags_df['person_id'].isin(persons_with_condition).astype(int)
        else:
            flags_df[col_name] = 0

    return flags_df

print("Batch processing function defined successfully!")
print("Ready to process batches.")
```

```python
# ============================================================================
# === CELL 6: Run Batch Processing with Checkpointing ===
# Main processing loop - can be resumed if interrupted
# ============================================================================

# Load batch assignments
all_persons = pd.read_csv(f"{OUTPUT_DIR}/batch_assignments.csv")
n_batches = all_persons['batch_number'].max()

# Check for existing checkpoint
checkpoint_file = f"{OUTPUT_DIR}/checkpoint.txt"
results_file = f"{RESULTS_DIR}/accumulated_results.csv"
conditions_file = f"{RESULTS_DIR}/accumulated_conditions.csv"

if os.path.exists(checkpoint_file):
    with open(checkpoint_file, 'r') as f:
        start_batch = int(f.read().strip()) + 1
    print(f"Resuming from batch {start_batch} (previous run completed batch {start_batch - 1})")

    # Load existing results
    if os.path.exists(results_file):
        accumulated_results = pd.read_csv(results_file)
    else:
        accumulated_results = pd.DataFrame()

    if os.path.exists(conditions_file):
        accumulated_conditions = pd.read_csv(conditions_file)
    else:
        accumulated_conditions = pd.DataFrame()
else:
    start_batch = 1
    accumulated_results = pd.DataFrame()
    accumulated_conditions = pd.DataFrame()
    print("Starting fresh from batch 1")

print("=" * 60)
print(f"PROCESSING BATCHES {start_batch} to {n_batches}")
print("=" * 60)

# Main processing loop
for batch_num in range(start_batch, n_batches + 1):
    print(f"\n--- Batch {batch_num}/{n_batches} ({100*batch_num/n_batches:.1f}%) ---")

    # Get person IDs for this batch
    batch_persons = all_persons[
        all_persons['batch_number'] == batch_num
    ]['person_id'].tolist()

    # Process batch
    demographics, conditions, success, error = process_batch(batch_num, batch_persons, CDR)

    if not success:
        print(f"ERROR: {error}")
        print("Progress saved. Re-run this cell to resume from this batch.")
        break

    # Create condition flags
    condition_flags = create_condition_flags(conditions, batch_persons)

    # Merge demographics with condition flags
    batch_results = demographics.merge(condition_flags, on='person_id', how='left')
    batch_results['batch_number'] = batch_num

    # Add to accumulated results
    accumulated_results = pd.concat([accumulated_results, batch_results], ignore_index=True)
    accumulated_conditions = pd.concat([accumulated_conditions, conditions], ignore_index=True)

    # Save checkpoint and intermediate results
    with open(checkpoint_file, 'w') as f:
        f.write(str(batch_num))

    accumulated_results.to_csv(results_file, index=False)
    accumulated_conditions.to_csv(conditions_file, index=False)

    # Save batch-specific file (optional, for debugging)
    batch_results.to_csv(f"{BATCH_DIR}/batch_{batch_num:03d}_results.csv", index=False)

    # Memory cleanup
    del demographics, conditions, condition_flags, batch_results
    gc.collect()

print("\n" + "=" * 60)
print("BATCH PROCESSING COMPLETE!")
print("=" * 60)
print(f"Total participants processed: {len(accumulated_results):,}")
print(f"Results saved to: {results_file}")
```

```python
# ============================================================================
# === CELL P7: Create PLINK-format Phenotype and Covariate Files ===
# ============================================================================

# Load phenotype data
phenotype_df = pd.read_csv(f"{RESULTS_DIR}/accumulated_results.csv")

# Create phenotype file for PLINK
# Format: FID IID pheno1 pheno2 ...
pheno_cols = ['has_asthma', 'has_hypertension', 'has_colorectal_cancer',
              'has_prostate_cancer', 'has_breast_cancer', 'has_atrial_fibrillation']

plink_pheno = phenotype_df[['person_id'] + pheno_cols].copy()
plink_pheno.insert(0, 'FID', plink_pheno['person_id'])
plink_pheno.rename(columns={'person_id': 'IID'}, inplace=True)

# PLINK uses -9 for missing, 1 for control, 2 for case (or 0/1 with --1)
# We'll use 0/1 format with --1 flag
plink_pheno.to_csv(f"{PLINK_DIR}/phenotypes.txt", sep='\t', index=False, na_rep='-9')
print(f"Phenotype file saved: {PLINK_DIR}/phenotypes.txt")

# Create covariate file
# Format: FID IID covar1 covar2 ...
covar_cols = ['age', 'gender']
if 'is_african_american' in phenotype_df.columns:
    covar_cols.append('is_african_american')

plink_covar = phenotype_df[['person_id'] + covar_cols].copy()
plink_covar.insert(0, 'FID', plink_covar['person_id'])
plink_covar.rename(columns={'person_id': 'IID'}, inplace=True)

plink_covar.to_csv(f"{PLINK_DIR}/covariates.txt", sep='\t', index=False, na_rep='-9')
print(f"Covariate file saved: {PLINK_DIR}/covariates.txt")

print("\nFiles ready for PLINK!")
print(f"  Phenotypes: {len(pheno_cols)} conditions")
print(f"  Covariates: {covar_cols}")
```

```python
# ============================================================================
# === CELL P8: After PLINK Extraction - Load and Process Results ===
# Run this AFTER the PLINK commands have completed
# ============================================================================

def load_plink_raw(raw_file):
    """
    Load PLINK --export A output (.raw file).

    The .raw file has columns:
    FID IID PAT MAT SEX PHENOTYPE SNP1_A SNP2_A ...

    SNP columns contain allele counts (0, 1, 2, or NA)
    """
    df = pd.read_csv(raw_file, sep='\s+')

    # Rename IID to person_id for merging
    df.rename(columns={'IID': 'person_id'}, inplace=True)

    # Drop PLINK metadata columns
    cols_to_drop = ['FID', 'PAT', 'MAT', 'SEX', 'PHENOTYPE']
    df.drop(columns=[c for c in cols_to_drop if c in df.columns], inplace=True)

    # Clean SNP column names (remove _A suffix that PLINK adds)
    df.columns = [c.replace('_A', '').replace('_C', '').replace('_G', '').replace('_T', '')
                  if c != 'person_id' else c for c in df.columns]

    return df

# Example usage (run after PLINK completes):
# genotypes_df = load_plink_raw(f"{PLINK_DIR}/genotypes_additive.raw")
# print(f"Loaded genotypes for {len(genotypes_df)} individuals")
# print(f"SNPs: {[c for c in genotypes_df.columns if c != 'person_id']}")
```

```python
# ============================================================================
# === CELL P9: Merge Genotypes with Phenotypes ===
# ============================================================================

def merge_and_prepare_analysis(phenotype_file, genotype_file, output_file):
    """
    Merge phenotype and genotype data for analysis.
    """
    # Load data
    pheno_df = pd.read_csv(phenotype_file)
    geno_df = load_plink_raw(genotype_file)

    # Merge on person_id
    merged = pheno_df.merge(geno_df, on='person_id', how='inner')

    print(f"Phenotype records: {len(pheno_df):,}")
    print(f"Genotype records: {len(geno_df):,}")
    print(f"Merged records: {len(merged):,}")

    # Check for missing genotypes
    snp_cols = [c for c in geno_df.columns if c != 'person_id']
    missing_rates = merged[snp_cols].isna().mean()
    print(f"\nMissing genotype rates:")
    for snp, rate in missing_rates.items():
        status = "⚠️ HIGH" if rate > 0.1 else "✓"
        print(f"  {snp}: {rate:.1%} {status}")

    # Save
    merged.to_csv(output_file, index=False)
    print(f"\nSaved merged data to: {output_file}")

    return merged

# Example usage (run after PLINK completes):
# analysis_df = merge_and_prepare_analysis(
#     f"{RESULTS_DIR}/accumulated_results.csv",
#     f"{PLINK_DIR}/genotypes_additive.raw",
#     f"{RESULTS_DIR}/merged_pheno_geno.csv"
# )
```

```python
# ============================================================================
# === CELL P10: Parse PLINK Association Results ===
# If you used PLINK for association testing
# ============================================================================

def load_plink_glm_results(results_dir, pattern="assoc_*.glm.logistic"):
    """
    Load PLINK2 --glm results files.
    """
    import glob

    all_results = []

    for filepath in glob.glob(f"{results_dir}/{pattern}"):
        # Extract phenotype name from filename
        pheno = os.path.basename(filepath).replace('assoc_', '').replace('.glm.logistic', '')

        df = pd.read_csv(filepath, sep='\t')
        df['phenotype'] = pheno
        all_results.append(df)

    if all_results:
        combined = pd.concat(all_results, ignore_index=True)
        return combined
    else:
        print(f"No results files found matching {pattern}")
        return pd.DataFrame()

# Example:
# plink_results = load_plink_glm_results(PLINK_DIR)
# print(plink_results.head())
```
